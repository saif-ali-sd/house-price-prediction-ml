```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_percentage_error
from sklearn import svm
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
import joblib

df_final = pd.read_csv("../data/processed/final_dataset.csv")

X = df_final.drop(['SalePrice'], axis=1)
Y = df_final['SalePrice']

X_train, X_valid, Y_train, Y_valid = train_test_split(
    X, Y, train_size=0.8, test_size=0.2, random_state=0)

# SVM
model_SVR = svm.SVR()
model_SVR.fit(X_train,Y_train)
Y_pred = model_SVR.predict(X_valid)
svr_mape = mean_absolute_percentage_error(Y_valid, Y_pred)

# Random Forest
model_RFR = RandomForestRegressor(n_estimators=10)
model_RFR.fit(X_train, Y_train)
rf_mape = mean_absolute_percentage_error(Y_valid, model_RFR.predict(X_valid))

# Linear Regression
model_LR = LinearRegression()
model_LR.fit(X_train, Y_train)
lr_mape = mean_absolute_percentage_error(Y_valid, model_LR.predict(X_valid))

print("SVR:", svr_mape)
print("RF:", rf_mape)
print("LR:", lr_mape)

joblib.dump(model_SVR, "../models/best_model.pkl")
print("Best model saved")
import pandas as pd

with open("../outputs/metrics.txt", "w") as f:
    f.write(f"SVR MAPE: {svr_mape}\n")
    f.write(f"Random Forest MAPE: {rf_mape}\n")
    f.write(f"Linear Regression MAPE: {lr_mape}\n")

sample = X_valid.head(10)
preds = model_SVR.predict(sample)

pd.DataFrame({
    "Actual": Y_valid.head(10),
    "Predicted": preds
}).to_csv("../outputs/sample_predictions.csv", index=False)

print("Outputs saved to outputs/")
```
